```python
import requests
import pprint as pp
```

## LightGCN

```python
# -----------------------------
# LightGCN 모델 테스트
# -----------------------------
url = "http://localhost:3000/predict_lightgcn"

# 입력: user_id 리스트 (추천 대상 유저)
payload = [{"user_id": 123}, {"user_id": 234}]

# REST API 호출 -> JSON 응답 수신
response = requests.post(url, json=payload)
response = response.json()

# 예측 결과 출력
pp.pprint(response)
```
